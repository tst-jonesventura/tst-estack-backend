```python
import os
import sys
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv())
backend_path = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.append(backend_path)
from src.database.database import create_session # noqa: E402
from src.database import models # noqa: E402
import nest_asyncio # noqa: E402
nest_asyncio.apply()
```

```python
with create_session() as session:
    kanban_column_list = session.query(models.ProjectKanbanColumn).all()
    print(f"Found {len(kanban_column_list)} kanban columns.\n")

    messages = {
        "Especificando": "Deseja iniciar a definição técnica e de negócio para esta história?",
        "Validando": "Atenção: Mover para Validando acionará a revisão de critérios de aceite. Prosseguir?",
        "Refinando": "Esta tarefa requer detalhamento técnico antes do desenvolvimento. Confirmar refinamento?",
        "Desenvolvendo": "Ao mover para cá, o tempo de ciclo (Lead Time) será iniciado. Confirmar início do Dev?",
        "Testando": "Certifique-se de que o ambiente de QA está pronto. Iniciar testes agora?",
        "Concluido": "Parabéns! Deseja finalizar esta tarefa e arquivar o histórico de alterações?"
    }

    for kanban_column in kanban_column_list:
        print(f"Processing kanban column: {kanban_column.id}-{kanban_column.name}")

        if kanban_column.column_change_warning is None:
            print(f"Setting column change warning for {kanban_column.name}")
            kanban_column.column_change_warning = messages.get(kanban_column.name)
            session.add(kanban_column)
        
        print(f"Finished processing kanban column: {kanban_column.id}-{kanban_column.name}\n")

    session.commit()
    print(f"Processed {len(kanban_column_list)} kanban columns.")
```
